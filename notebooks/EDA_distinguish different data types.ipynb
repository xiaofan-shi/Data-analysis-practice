```python
import pandas as pd
```

```python
df=pd.read_csv('Leads.csv')
```

```python
df.columns #找具体栏目
```

```python
continuous_candidates = df.select_dtypes(include=['int64', 'float64']).columns
categorical_candidates = df.select_dtypes(include=['object']).columns

# To explore the df first, we decide to categorize the data, we could better Know what to focus and discussed!
categorical_numeric = [
    'Prospect ID', 'Lead Number', 'Lead Origin', 'Lead Source',
       'Do Not Email', 'Do Not Call', 'Converted', 'TotalVisits',
       'Total Time Spent on Website', 'Page Views Per Visit', 'Last Activity',
       'Country'
]

# Finalize variable classification
continuous_vars = [col for col in continuous_candidates if col not in categorical_numeric]
categorical_vars = list(categorical_candidates) + categorical_numeric

# 5. Calculate and print results
print("\n" + "="*50)
print("=== Variable Type Statistics ===")
print(f"1. Number of Continuous Variables: {len(continuous_vars)}")
print(f"   Specific variables: {continuous_vars}")
print(f"\n2. Number of Categorical Variables: {len(categorical_vars)}")
print(f"   Specific variables: {categorical_vars}")
```
